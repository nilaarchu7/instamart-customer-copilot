```python
query6 = """
WITH customer_spending AS (
    SELECT
        c.customer_id,
        c.city,
        SUM(o.order_amount) AS total_spending
    FROM customers c
    JOIN orders o
        ON c.customer_id = o.customer_id
    WHERE o.status = 'completed'
    GROUP BY c.customer_id, c.city
)

SELECT
    customer_id,
    city,
    total_spending,

    RANK() OVER (
        PARTITION BY city
        ORDER BY total_spending DESC
    ) AS rank_position,

    DENSE_RANK() OVER (
        PARTITION BY city
        ORDER BY total_spending DESC
    ) AS dense_rank_position,

    ROW_NUMBER() OVER (
        PARTITION BY city
        ORDER BY total_spending DESC
    ) AS row_number_position

FROM customer_spending
ORDER BY city, total_spending DESC;
"""

result6 = pd.read_sql_query(query6, conn)

print(result6.head(15))
print("Total customers:", len(result6))
```

```python
import pandas as pd

customers = pd.read_csv("../data/clean/customers_clean.csv")
orders = pd.read_csv("../data/clean/orders_clean.csv")

print("Customers shape:", customers.shape)
print("Orders shape:", orders.shape)

print("\nCustomer columns:")
print(customers.columns.tolist())

print("\nOrder columns:")
print(orders.columns.tolist())
```

```python
completed_orders = orders[orders["status"] == "completed"].copy()

print("Completed orders:", len(completed_orders))
print(completed_orders.head())
```

```python
completed_orders["order_ts"] = pd.to_datetime(completed_orders["order_ts"])

snapshot_date = pd.Timestamp("2025-10-01 00:00:00")

print("Date conversion completed!")
print("Latest order date:", completed_orders["order_ts"].max())
print("Snapshot date:", snapshot_date)
```

```python
rfm = completed_orders.groupby("customer_id").agg(
    last_order_date=("order_ts", "max"),
    frequency=("order_id", "nunique"),
    monetary=("order_amount", "sum")
).reset_index()

rfm["recency"] = (
    snapshot_date - rfm["last_order_date"]
).dt.days

rfm = rfm[
    ["customer_id", "recency", "frequency", "monetary"]
]

print(rfm.head(10))
print("\nTotal customers in RFM:", len(rfm))
```

```python
print("RFM Summary:")
print(rfm[["recency", "frequency", "monetary"]].describe())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

rfm_scaled = scaler.fit_transform(
    rfm[["recency", "frequency", "monetary"]]
)

print("Original RFM shape:", rfm.shape)
print("Standardized RFM shape:", rfm_scaled.shape)

print("\nFirst 5 standardized rows:")
print(rfm_scaled[:5])
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

results = []

for k in range(2, 9):

    kmeans = KMeans(
        n_clusters=k,
        n_init=10,
        random_state=42
    )

    labels = kmeans.fit_predict(rfm_scaled)

    inertia = kmeans.inertia_
    silhouette = silhouette_score(rfm_scaled, labels)

    results.append({
        "K": k,
        "Inertia": inertia,
        "Silhouette": silhouette
    })

results_df = pd.DataFrame(results)

print(results_df.round(3))
```

```python
final_kmeans = KMeans(
    n_clusters=4,
    n_init=10,
    random_state=42
)

rfm["segment_id"] = final_kmeans.fit_predict(rfm_scaled)

segment_profile = rfm.groupby("segment_id").agg(
    customer_count=("customer_id", "count"),
    avg_recency=("recency", "mean"),
    avg_frequency=("frequency", "mean"),
    avg_monetary=("monetary", "mean")
).round(2)

print(segment_profile)
```

```python
segment_names = {
    0: "Premium Customers",
    1: "Occasional Customers",
    2: "At-Risk Customers",
    3: "Loyal Customers"
}

rfm["segment_name"] = rfm["segment_id"].map(segment_names)

# Find the segment with the highest average spending
premium_segment = segment_profile["avg_monetary"].idxmax()

rfm["is_premium"] = rfm["segment_id"] == premium_segment

print(rfm.head(10))

print("\nPremium segment ID:", premium_segment)
print("Number of Premium customers:", rfm["is_premium"].sum())
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)

rfm_pca = pca.fit_transform(rfm_scaled)

print("Explained variance ratio:")
print(pca.explained_variance_ratio_)

print("\nTotal variance explained:")
print(pca.explained_variance_ratio_.sum())
```

```python
# Convert order dates into datetime format
orders["order_ts"] = pd.to_datetime(orders["order_ts"])

# Define the cutoff date
cutoff_date = pd.Timestamp("2025-09-01")

# Find completed orders before September 1
before_cutoff = orders[
    (orders["status"] == "completed") &
    (orders["order_ts"] < cutoff_date)
]

# Find completed orders during September
september_orders = orders[
    (orders["status"] == "completed") &
    (orders["order_ts"] >= cutoff_date) &
    (orders["order_ts"] < pd.Timestamp("2025-10-01"))
]

# Customers eligible for lapse prediction
cohort = pd.DataFrame({
    "customer_id": before_cutoff["customer_id"].unique()
})

# Mark customers who ordered during September
september_customer_ids = set(september_orders["customer_id"])

cohort["lapsed"] = cohort["customer_id"].apply(
    lambda x: 0 if x in september_customer_ids else 1
)

print("Total eligible customers:", len(cohort))
print("\nLapse counts:")
print(cohort["lapsed"].value_counts().sort_index())

print("\nLapsed percentage:")
print(round(cohort["lapsed"].mean() * 100, 2), "%")
```

```python
# Convert signup dates
customers["signup_date"] = pd.to_datetime(customers["signup_date"])

# All orders before September 1 (completed + cancelled)
orders_before_cutoff = orders[
    orders["order_ts"] < cutoff_date
].copy()

# Completed orders before September 1
completed_before_cutoff = orders_before_cutoff[
    orders_before_cutoff["status"] == "completed"
].copy()

# Add product category to completed orders
completed_before_cutoff = completed_before_cutoff.merge(
    products[["product_id", "category"]],
    on="product_id",
    how="left"
)

print("All orders before cutoff:", len(orders_before_cutoff))
print("Completed orders before cutoff:", len(completed_before_cutoff))

print("\nProduct categories:")
print(completed_before_cutoff["category"].value_counts())
```

```python
completed_features = completed_before_cutoff.groupby("customer_id").agg(
    last_order_date=("order_ts", "max"),
    frequency=("order_id", "nunique"),
    monetary=("order_amount", "sum"),
    avg_discount=("discount_pct", "mean"),
    avg_delivery=("delivery_minutes", "mean")
).reset_index()

# Calculate recency using September 1 as cutoff
completed_features["recency"] = (
    cutoff_date - completed_features["last_order_date"]
).dt.days

# Remove the last order date column
completed_features = completed_features.drop(
    columns=["last_order_date"]
)

print(completed_features.head())
print("\nTotal customers:", len(completed_features))
```

```python
cancellation_features = orders_before_cutoff.groupby("customer_id").agg(
    total_orders=("order_id", "nunique"),
    cancelled_orders=("status", lambda x: (x == "cancelled").sum())
).reset_index()

cancellation_features["cancellation_rate"] = (
    cancellation_features["cancelled_orders"] /
    cancellation_features["total_orders"]
)

print(cancellation_features.head())
```

```python
favourite_category = (
    completed_before_cutoff
    .groupby(["customer_id", "category"])
    .size()
    .reset_index(name="category_count")
)

# Select the most frequently purchased category per customer
favourite_category = (
    favourite_category
    .sort_values(
        ["customer_id", "category_count", "category"],
        ascending=[True, False, True]
    )
    .drop_duplicates("customer_id")
)

favourite_category = favourite_category[
    ["customer_id", "category"]
]

favourite_category = favourite_category.rename(
    columns={"category": "favourite_category"}
)

print(favourite_category.head(10))
print("\nTotal customers:", len(favourite_category))
```

```python
# Start with completed-order features
features = completed_features.copy()

# Add cancellation rate
features = features.merge(
    cancellation_features[
        ["customer_id", "cancellation_rate"]
    ],
    on="customer_id",
    how="left"
)

# Add favourite category
features = features.merge(
    favourite_category,
    on="customer_id",
    how="left"
)

# Add customer details
features = features.merge(
    customers[
        ["customer_id", "city", "age", "gender", "signup_date"]
    ],
    on="customer_id",
    how="left"
)

# Calculate customer tenure
features["tenure_days"] = (
    cutoff_date - features["signup_date"]
).dt.days

# Add lapse labels
features = features.merge(
    cohort[["customer_id", "lapsed"]],
    on="customer_id",
    how="inner"
)

# Remove signup date because tenure is already calculated
features = features.drop(columns=["signup_date"])

print("Feature table shape:", features.shape)
print("\nFeature columns:")
print(features.columns.tolist())

print("\nFirst 5 rows:")
print(features.head())
```

```python
import re

# Create a separate working copy
tickets_clean = tickets.copy()

# Function to hide personal information
def redact_pii(text):

    text = str(text)

    # Hide email addresses
    text = re.sub(
        r'\b[\w.-]+@[\w.-]+\.\w+\b',
        '[EMAIL]',
        text
    )

    # Hide order IDs
    text = re.sub(
        r'\bORD-\d+\b',
        '[ORDER_ID]',
        text,
        flags=re.IGNORECASE
    )

    # Hide phone numbers
    text = re.sub(
        r'(?<!\w)(?:\+?\d{1,3}[-.\s]?)?(?:\d[-.\s()]?){9,12}(?!\w)',
        '[PHONE]',
        text
    )

    return text


# Apply redaction
tickets_clean["safe_text"] = tickets_clean["text"].fillna("").apply(redact_pii)

# Assign processing status
tickets_clean["processing_status"] = "ready"

tickets_clean.loc[
    tickets_clean["text_length"] == 0,
    "processing_status"
] = "skip_empty"

tickets_clean.loc[
    tickets_clean["text_length"] > 2000,
    "processing_status"
] = "human_review"


# Display results
print("Processing status:")
print(tickets_clean["processing_status"].value_counts())

print("\nOriginal ticket:")
print(tickets_clean["text"].iloc[0])

print("\nRedacted ticket:")
print(tickets_clean["safe_text"].iloc[0])
```

```python
def call_llm_with_retry(messages, max_retries=3):

    for attempt in range(max_retries):

        try:
            response = mock_llm.chat(messages)
            return response

        except mock_llm.TransientLLMError as e:

            print(f"Temporary error: {e}")
            print(f"Retry attempt: {attempt + 1}")

    return None
```

```python
import json

def validate_response(response):

    if response is None:
        return None

    try:
        result = json.loads(response.choices[0].message.content)

        if not isinstance(result, dict):
            return None

        category = result.get("category")

        if category not in mock_llm.CATS:
            return None

        return result

    except (json.JSONDecodeError, TypeError, AttributeError):
        return None
```

```python
zero_shot_prompt = """
CLASSIFY

Classify the customer support ticket into exactly one category:
billing, delivery, technical, or fraud.

Return only valid JSON in this format:
{"category": "category_name", "reason": "short reason"}
"""

few_shot_prompt = """
CLASSIFY

Classify the customer support ticket into exactly one category:
billing, delivery, technical, or fraud.

EXAMPLES:
Ticket: I was charged twice for my order.
Category: billing

Ticket: My delivery is late.
Category: delivery

Ticket: The app keeps crashing.
Category: technical

Ticket: Someone hacked my account.
Category: fraud

Return only valid JSON in this format:
{"category": "category_name", "reason": "short reason"}
"""

print("Zero-shot prompt created.")
print("\nFew-shot prompt created.")
```

```python
def classify_ticket(ticket_text, prompt):

    messages = [
        {
            "role": "system",
            "content": prompt
        },
        {
            "role": "user",
            "content": ticket_text
        }
    ]

    # Call LLM with retry
    response = call_llm_with_retry(messages)

    # Validate the response
    result = validate_response(response)

    if result is None:
        return {
            "predicted_category": None,
            "status": "invalid_response"
        }

    return {
        "predicted_category": result["category"],
        "status": "success"
    }


print("Ticket classification function created successfully!")
```

```python
import random
import time

BASE_DELAY = 0.05
MAX_RETRIES = 3

def call_llm_with_retry(messages, ticket_id):
    for attempt in range(MAX_RETRIES + 1):
        try:
            response = mock_llm.chat(messages)
            return response

        except mock_llm.TransientLLMError:

            # Stop if all 3 retries are exhausted
            if attempt == MAX_RETRIES:
                print(f"Ticket {ticket_id}: all retries exhausted")
                raise

            # Exponential backoff with jitter
            wait_time = (
                BASE_DELAY
                * (2 ** attempt)
                * random.uniform(0.9, 1.15)
            )

            print(
                f"Ticket {ticket_id} | "
                f"Retry {attempt} | "
                f"Wait: {wait_time:.4f} seconds"
            )

            time.sleep(wait_time)

print("Retry function created successfully.")
```

```python
import json
import re

ALLOWED_CATEGORIES = ["billing", "delivery", "technical", "fraud"]

def validate_response(response_text):

    # First: Try JSON validation
    try:
        data = json.loads(response_text)
        category = data.get("category", "").lower()

        if category in ALLOWED_CATEGORIES:
            return category, "classified_json"

    except (json.JSONDecodeError, AttributeError):
        pass

    # Second: Try keyword extraction
    words = re.findall(r"[a-z]+", response_text.lower())

    found_categories = [
        word for word in words
        if word in ALLOWED_CATEGORIES
    ]

    if len(found_categories) == 1:
        return found_categories[0], "classified_keyword"

    # Otherwise: Human review
    return None, "flagged_invalid_output"


print("Response validation function created successfully.")
```

```python
def run_pipeline(prompt, input_tickets, token_budget=None):

    results = []
    tokens_used = 0

    for _, row in input_tickets.iterrows():

        ticket_id = row["ticket_id"]
        safe_text = row["safe_text"]

        if token_budget is not None and tokens_used >= token_budget:
            results.append({
                "ticket_id": ticket_id,
                "true_category": row["true_category"],
                "category": None,
                "status": "budget_exhausted",
                "tokens_used": 0,
                "segment_name": row["segment_name"],
                "is_premium": row["is_premium"]
            })
            continue

        messages = [
            {"role": "system", "content": prompt},
            {"role": "user", "content": safe_text}
        ]

        # PII safety checks
        assert messages[1]["content"] == safe_text
        assert not re.search(r'[\w.+-]+@[\w-]+\.[\w.]+', safe_text)
        assert not re.search(r'\b[6-9]\d{9}\b', safe_text)
        assert not re.search(r'ORD-\d{6}', safe_text)

        try:
            response = call_llm_with_retry(messages, ticket_id)

            response_text = response.choices[0].message.content
            ticket_tokens = response.usage.total_tokens

            tokens_used += ticket_tokens

            category, status = validate_response(response_text)

        except mock_llm.TransientLLMError:
            category = None
            status = "flagged_invalid_output"
            ticket_tokens = 0

        results.append({
            "ticket_id": ticket_id,
            "true_category": row["true_category"],
            "category": category,
            "status": status,
            "tokens_used": ticket_tokens,
            "segment_name": row["segment_name"],
            "is_premium": row["is_premium"]
        })

    return pd.DataFrame(results)


# Reset immediately before the full zero-shot run
mock_llm.reset()

zero_shot_results = run_pipeline(
    zero_shot_prompt,
    ready_tickets
)

print("Zero-shot pipeline completed.")
print("\nStatus counts:")
print(zero_shot_results["status"].value_counts())

print("\nTotal tokens used:", zero_shot_results["tokens_used"].sum())
print("Total tickets processed:", len(zero_shot_results))
```

```python
import os

policy_folder = "../data/policies"

policy_chunks = []

chunk_size = 40
overlap = 10
step = chunk_size - overlap

for filename in os.listdir(policy_folder):
    if filename.endswith(".txt"):
        file_path = os.path.join(policy_folder, filename)

        with open(file_path, "r", encoding="utf-8") as file:
            text = file.read()

        words = text.split()

        for start in range(0, len(words), step):
            chunk_words = words[start:start + chunk_size]

            if chunk_words:
                policy_chunks.append({
                    "source": filename,
                    "text": " ".join(chunk_words)
                })

            if start + chunk_size >= len(words):
                break

print("Total policy chunks:", len(policy_chunks))
```

```python
import re
from collections import Counter

def retrieve_policy(question, top_k=3):
    question_words = set(re.findall(r"\b[a-zA-Z]+\b", question.lower()))

    results = []

    for chunk in policy_chunks:
        chunk_words = set(re.findall(r"\b[a-zA-Z]+\b", chunk["text"].lower()))

        score = len(question_words & chunk_words)

        if score > 0:
            results.append({
                "source": chunk["source"],
                "text": chunk["text"],
                "score": score
            })

    results.sort(key=lambda x: x["score"], reverse=True)

    return results[:top_k]


question = "Can I get a refund for a late delivery?"

retrieved_chunks = retrieve_policy(question)

for i, chunk in enumerate(retrieved_chunks):
    print(f"\nResult {i + 1}")
    print("Source:", chunk["source"])
    print("Score:", chunk["score"])
    print("Text:", chunk["text"])
```

```python
import re

stop_words = {
    "a", "an", "the", "is", "are", "can", "i",
    "for", "to", "of", "in", "on", "my", "do",
    "get", "what", "how", "when", "if", "and"
}

def retrieve_policy(question, top_k=3):
    question_words = set(re.findall(r"\b[a-zA-Z]+\b", question.lower()))
    question_words = question_words - stop_words

    results = []

    for chunk in policy_chunks:
        chunk_words = set(re.findall(r"\b[a-zA-Z]+\b", chunk["text"].lower()))

        score = len(question_words & chunk_words)

        if score > 0:
            results.append({
                "source": chunk["source"],
                "text": chunk["text"],
                "score": score
            })

    results.sort(key=lambda x: x["score"], reverse=True)

    return results[:top_k]


question = "Can I get a refund for a late delivery?"

retrieved_chunks = retrieve_policy(question)

for i, chunk in enumerate(retrieved_chunks):
    print(f"\nResult {i + 1}")
    print("Source:", chunk["source"])
    print("Score:", chunk["score"])
    print("Text:", chunk["text"])
```

```python
evaluation_results = []

mock_llm.reset()

for _, row in eval_df.iterrows():

    question = row["question"]
    expected_doc = row["expected_doc"]
    expected_fact = row["expected_fact"]

    # Retrieve relevant policy chunks
    retrieved = retrieve_policy(question, top_k=3)

    context = "\n\n".join(
        chunk["text"] for chunk in retrieved
    )

    # Ask the policy assistant
    messages = [
        {
            "role": "system",
            "content": "You are a POLICY ASSISTANT. Answer using only the provided policy context. If the answer is not available, say: I could not find that information in the policy documents."
        },
        {
            "role": "user",
            "content": f"CONTEXT:\n{context}\n\nQUESTION:\n{question}"
        }
    ]

    response = mock_llm.chat(messages)
    answer = response.choices[0].message.content

    retrieved_doc = retrieved[0]["source"] if retrieved else "NONE"

    # Check whether the expected document was retrieved
    doc_match = (
        retrieved_doc == expected_doc
        if expected_doc != "NONE"
        else retrieved_doc == "NONE"
    )

    # Check whether the expected fact appears in the answer
    if expected_doc == "NONE":
        fact_match = "could not find" in answer.lower()
    else:
        fact_match = (
            str(expected_fact).lower() in answer.lower()
        )

    evaluation_results.append({
        "question": question,
        "expected_doc": expected_doc,
        "retrieved_doc": retrieved_doc,
        "answer": answer,
        "doc_match": doc_match,
        "fact_match": fact_match
    })

results_df = pd.DataFrame(evaluation_results)

print("Total questions:", len(results_df))
print("Document retrieval matches:", results_df["doc_match"].sum())
print("Expected facts found:", results_df["fact_match"].sum())

display(results_df)
```

```python
def get_order(order_id):
    order = orders_df[orders_df["order_id"] == order_id]

    if order.empty:
        return {
            "found": False,
            "message": "Order not found"
        }

    return {
        "found": True,
        "order": order.iloc[0].to_dict()
    }


# Test the tool
print(get_order("ORD-100048"))
print(get_order("ORD-999999"))
```

```python
def search_policy(query):
    results = retrieve_policy(query, top_k=3)

    if not results:
        return "No relevant policy information found."

    return "\n\n".join(
        f"Source: {item['source']}\n{item['text']}"
        for item in results
    )


# Test the tool
print(search_policy("What is the refund rule for late delivery?"))
```

```python
def calculate_refund(order):
    if order["status"] != "completed":
        return {
            "eligible": False,
            "refund_amount": 0,
            "reason": "Only completed orders can be refunded."
        }

    if float(order["delivery_minutes"]) <= 45:
        return {
            "eligible": False,
            "refund_amount": 0,
            "reason": "Delivery was within 45 minutes."
        }

    return {
        "eligible": True,
        "refund_amount": float(order["order_amount"]),
        "reason": "Eligible for late-delivery refund."
    }


# Test the tool
order_details = get_order("ORD-100048")

if order_details["found"]:
    print(calculate_refund(order_details["order"]))
```

```python
import json

refund_records = []
escalation_records = []

# Tool 1: Get order details
def get_order(order_id):
    order = orders_df[orders_df["order_id"] == order_id]

    if order.empty:
        return {"error": "Order not found"}

    return order.iloc[0].to_dict()


# Tool 2: Search policy
def search_policy(query):
    results = retrieve_policy(query, top_k=3)

    return {
        "policy": "\n\n".join(
            item["text"] for item in results
        )
    }


# Tool 3: Calculate refund
def calculate_refund(order_amount, delivery_minutes):
    if float(delivery_minutes) > 45:
        return {"refund_inr": float(order_amount)}

    return {"refund_inr": 0}


# Tool 4: Issue refund
def issue_refund(order_id, refund_inr):
    record = {
        "order_id": order_id,
        "refund_inr": float(refund_inr),
        "status": "REFUND_ISSUED"
    }

    refund_records.append(record)
    return record


# Tool 5: Escalate to human
def escalate_to_human(order_id, refund_inr):
    record = {
        "order_id": order_id,
        "refund_inr": float(refund_inr),
        "status": "PENDING_HUMAN_APPROVAL"
    }

    escalation_records.append(record)
    return record


print("All 5 tools updated successfully!")
```

```python
import json
import mock_llm

# Connect tool names to Python functions
tool_functions = {
    "get_order": get_order,
    "search_policy": search_policy,
    "calculate_refund": calculate_refund,
    "issue_refund": issue_refund,
    "escalate_to_human": escalate_to_human
}


def run_agent(user_query, max_steps=6):

    messages = [
        {
            "role": "system",
            "content": "You are an AGENT. Help customers with refund requests using the available tools."
        },
        {
            "role": "user",
            "content": user_query
        }
    ]

    for step in range(max_steps):

        response = mock_llm.chat(messages)

        message = response.choices[0].message
        tool_calls = message.tool_calls

        # If there are no more tool calls, return the final answer
        if not tool_calls:
            return message.content

        # Execute the requested tool
        for tool_call in tool_calls:

            tool_name = tool_call.name
            arguments = tool_call.arguments

            if isinstance(arguments, str):
                arguments = json.loads(arguments)

            print(f"Step {step + 1}: Calling {tool_name}")

            if tool_name not in tool_functions:
                result = {"error": "Unknown tool"}
            else:
                result = tool_functions[tool_name](**arguments)

            # Send tool result back to the LLM
            messages.append({
                "role": "tool",
                "name": tool_name,
                "content": json.dumps(result),
                "tool_call_id": tool_call.id
            })

    return "FALLBACK: Maximum tool steps reached."


print("ReAct agent created successfully!")
```

```python
mock_llm.reset()

agent_results = []

for _, row in scenarios_df.iterrows():

    print("\n" + "=" * 50)
    print("Scenario:", row["scenario_id"])

    result = run_agent(row["message"])

    print("Expected:", row["expected_outcome"])
    print("Actual:", result)

    # Identify actual outcome
    actual_outcome = result.split(":")[0].strip()

    agent_results.append({
        "scenario_id": row["scenario_id"],
        "expected_outcome": row["expected_outcome"],
        "actual_outcome": actual_outcome,
        "result": result,
        "passed": actual_outcome == row["expected_outcome"]
    })

agent_results_df = pd.DataFrame(agent_results)

print("\n\nFINAL AGENT RESULTS")
display(agent_results_df)

print(
    "Scenarios passed:",
    agent_results_df["passed"].sum(),
    "/",
    len(agent_results_df)
)
```
